```python
#CLASSE FAMÍLIA
class Familia:

    def __init__(self, repre_familiar, conjugue, qtd_filhos, renda, formacao, qtd_banheiros, qtd_tvs, net):
        self.repre_familiar = repre_familiar
        self.conjugue = conjugue
        self.qtd_filhos = qtd_filhos
        self.renda = renda
        self.formacao = formacao
        self.qtd_banheiros = qtd_banheiros
        self.qtd_tvs = qtd_tvs
        self.net = net

    #SEU CÓDIGO ....

    #MÉTODO USADO PARA CRIAR O GRAFO PARA FAMILIA
    def criar_grafo_familia(self):
        filhos = []
        banheiros = []
        tvs = []

        if self.qtd_filhos > 0:
            filhos = [ f"F{(x+1)}" for x in range(self.qtd_filhos)]

        if self.qtd_banheiros > 0:
            banheiros = [ f"B{(x+1)}" for x in range(self.qtd_banheiros)]

        if self.qtd_tvs > 0:
            tvs = [ f"Tv{(x+1)}" for x in range(self.qtd_tvs) ]

        grafo = {
            "FA":["RF","F","CJ","RE"],
            "RF":[self.repre_familiar],
            "CJ":[self.conjugue],
            "F":filhos,
            "RE":["Renda","B","Tv","Net"],
            "Renda":[str(self.renda)],
            "B":banheiros,
            "Tv":tvs,
            "Net":[self.net]
        }

        for f in filhos:
            grafo[f] = []

        for b in banheiros:
            grafo[b] = []

        for t in tvs :
            grafo[t] = []

        grafo[self.net] = []
        grafo[self.repre_familiar] = []
        grafo[self.conjugue] = []
        grafo[str(self.renda)]= []

        return grafo
```

```python
class Censo:
    def __init__(self):
        self.registros = []
        self.algoritmo = Algoritmo()

    #GRAFO PARA PERCORRER AS CIDADES
    grafo_cidades = {
        "SERRA":["VITORIA","ARACRUZ"],
        "VITORIA": ["VILAVELHA","CARIACICA"],
        "VILAVELHA":["VIANA","CARIACICA","GUARAPARI"],
        "GUARAPARI":[],
        "CARIACICA":["VIANA"],
        "VIANA":[],
        "ARACRUZ":[],
    }
    #MÉTODO PARA GERAR OS REGISTROS PARA O CENSO (VISISTA PORTA A PORTA)
    def gerar_registros(self, ordem_visita, familias_cidade):

        for cidade in ordem_visita:
            for familia in familias_cidade.dados_familias[cidade]:
                self.registros.append( {
                    'cidade': cidade,
                    'representante familiar':familia.repre_familiar,
                    'conjugue':familia.conjugue,
                    'renda': familia.renda,
                    'formacao':familia.formacao,
                    'banheiro':familia.qtd_banheiros,
                    'tv':familia.qtd_tvs,
                    'internet':familia.net
                })
        return self.registros

    #MÉTODO PARA OBTER INFORMAÇÕES DAS FAMILIAS ATRÁVES DO GRAFO, USANDO DFS
    def info_familia(self,grafo, inicio = "RF"):
        return self.algoritmo.dfs(grafo, inicio)

    #MÉTODO PARA OBTER A ORDEM DE VISITA USANDO O BFS
    def ordem_visita(self):
        return self.algoritmo.bfs(self.grafo_cidades,"SERRA")
```

```python
from collections import deque

class Algoritmo:

    def bfs(self,grafo ,inicio):
        visitados = set()
        queue = deque([inicio])
        ordem = []

        while queue:
            node = queue.popleft()
            ordem.append(node)

            for vizinho in grafo[node]:
                if vizinho not in visitados:
                    visitados.add(vizinho)
                    queue.append(vizinho)

        return ordem

    def dfs(self, grafo ,inicio, visitados = None, ordem = None):
        if visitados is None:
            visitados = set()
        if ordem is None:
            ordem = []

        visitados.add(inicio)
        ordem.append(inicio)

        for vizinho in grafo[inicio]:
            if vizinho not in visitados:
                visitados.add(vizinho)
                self.dfs(grafo, vizinho, visitados, ordem)

        return ordem
```
