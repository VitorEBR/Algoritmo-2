```python
#CLASSE FAMÍLIA
class Familia:

    #SEU CÓDIGO ....

    #MÉTODO USADO PARA CRIAR O GRAFO PARA FAMILIA
    def criar_grafo_familia(self):
        filhos = []
        banheiros = []
        tvs = []

        if self.qtd_filhos > 0:
            filhos = [ f"F{(x+1)}" for x in range(self.qtd_filhos)]

        if self.qtd_banheiros > 0:
            banheiros = [ f"B{(x+1)}" for x in range(self.qtd_banheiros)]

        if self.qtd_tvs > 0:
            tvs = [ f"Tv{(x+1)}" for x in range(self.qtd_tvs) ]

        grafo = {
            "FA":["RF","F","CJ","RE"],
            "RF":[self.repre_familiar],
            "CJ":[self.conjugue],
            "F":filhos,
            "RE":["Renda","B","Tv","Net"],
            "Renda":[str(self.renda)],
            "B":banheiros,
            "Tv":tvs,
            "Net":[self.net]
        }

        for f in filhos:
            grafo[f] = []

        for b in banheiros:
            grafo[b] = []

        for t in tvs :
            grafo[t] = []

        grafo[self.net] = []
        grafo[self.repre_familiar] = []
        grafo[self.conjugue] = []
        grafo[str(self.renda)]= []

        return grafo
```

```python
from collections import deque

class Algoritmo:

    def bfs(self,grafo ,inicio):
        visitados = set()
        queue = deque([inicio])
        ordem = []

        while queue:
            node = queue.popleft()
            ordem.append(node)

            for vizinho in grafo[node]:
                if vizinho not in visitados:
                    visitados.add(vizinho)
                    queue.append(vizinho)

        return ordem

    def dfs(self, grafo ,inicio, visitados = None, ordem = None):
        if visitados is None:
            visitados = set()
        if ordem is None:
            ordem = []

        visitados.add(inicio)
        ordem.append(inicio)

        for vizinho in grafo[inicio]:
            if vizinho not in visitados:
                visitados.add(vizinho)
                self.dfs(grafo, vizinho, visitados, ordem)

        return ordem
```
